```python
import os, glob, math, random, warnings
import numpy as np, pandas as pd
import torch, torch.nn as nn, torchvision
from PIL import Image
from tqdm.auto import tqdm
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore")

SEED = 42
def seed_all(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s); torch.cuda.manual_seed_all(s)
seed_all(SEED)

dev = "cuda"
assert torch.cuda.is_available(), "Turn on a GPU: Runtime > Change runtime type > T4"

paths = glob.glob("data/**/*.jpg", recursive=True)
assert len(paths) > 0, "No images found. Run Cell 0 first and check the cwd is /content"

def parse(p):
    s = p.split(os.sep)
    i = s.index("SOB")                        # .../<class>/SOB/<subtype>/<slide>/<mag>/file
    return dict(path=p,
                label=int(s[i - 1] == "malignant"),
                subtype=s[i + 1],
                patient=s[i + 2].split("_")[-1],
                mag=int(s[i + 3].rstrip("X")))

df = pd.DataFrame([parse(p) for p in paths]).sort_values("path").reset_index(drop=True)

print(f"{len(df)} images | {df.patient.nunique()} patients (expected 7909 / 82)")
print(df.groupby("mag").label.value_counts().unstack().rename(columns={0: "benign", 1: "malignant"}))
pp = df.groupby("patient").label.first()
print("patients:", (pp == 0).sum(), "benign,", (pp == 1).sum(), "malignant")
```

```python
def load(p):
    im = Image.open(p).convert("RGB")
    if im.size != (224, 224):
        im = im.resize((224, 224), Image.BILINEAR)
    return np.asarray(im)

X = torch.from_numpy(np.stack([load(p) for p in tqdm(df.path)])).to(dev)  # N,H,W,3 uint8
Y = torch.tensor(df.label.values, device=dev)
y = df.label.values
print(X.shape, X.dtype)
```

```python
MEAN = torch.tensor([0.485, 0.456, 0.406], device=dev).view(1, 3, 1, 1)
STD  = torch.tensor([0.229, 0.224, 0.225], device=dev).view(1, 3, 1, 1)
EPOCHS, BS, LR = 8, 64, 3e-4        # set EPOCHS=3 for a quick test run

def prep(idx, train):
    x = X[idx].permute(0, 3, 1, 2).float() / 255
    if train:
        B = x.size(0)
        for dim in (3, 2):                                   # random h/v flips
            m = (torch.rand(B, device=dev) < 0.5).view(B, 1, 1, 1)
            x = torch.where(m, x.flip(dim), x)
        x = torch.rot90(x, random.randint(0, 3), (2, 3))     # random 90-degree rotation
        gain = 1 + (torch.rand(B, 3, 1, 1, device=dev) - 0.5) * 0.2   # per-channel gain (stain variation)
        c = 1 + (torch.rand(B, 1, 1, 1, device=dev) - 0.5) * 0.3      # contrast
        mu = x.mean((1, 2, 3), keepdim=True)
        x = ((x - mu) * c + mu) * gain
        x = x.clamp(0, 1)
    return (x - MEAN) / STD

def make_model():
    m = torchvision.models.resnet18(weights="IMAGENET1K_V1")
    m.fc = nn.Sequential(nn.Dropout(0.3), nn.Linear(m.fc.in_features, 2))
    return m.to(dev).to(memory_format=torch.channels_last)

@torch.no_grad()
def predict(model, idx):
    model.eval(); out = []
    for i in range(0, len(idx), 256):
        b = torch.as_tensor(idx[i:i+256], device=dev)
        with torch.autocast("cuda", dtype=torch.float16):
            out.append(torch.softmax(model(prep(b, False)).float(), 1)[:, 1].cpu())
    return torch.cat(out).numpy()

def run_fold(tr, te, seed=0):
    seed_all(seed)
    model = make_model()
    cw = torch.tensor(len(tr) / (2 * np.bincount(y[tr])), dtype=torch.float32, device=dev)
    crit = nn.CrossEntropyLoss(weight=cw, label_smoothing=0.1)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
    steps = EPOCHS * math.ceil(len(tr) / BS)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=LR, total_steps=steps, pct_start=0.2)
    scaler = torch.amp.GradScaler("cuda")
    for ep in range(EPOCHS):
        model.train()
        perm = np.random.permutation(tr)
        for i in range(0, len(perm), BS):
            b = torch.as_tensor(perm[i:i+BS], device=dev)
            if len(b) < 8: continue
            with torch.autocast("cuda", dtype=torch.float16):
                loss = crit(model(prep(b, True)), Y[b])
            opt.zero_grad(set_to_none=True)
            scaler.scale(loss).backward(); scaler.step(opt); scaler.update(); sched.step()
    return predict(model, te)   # fixed epochs, no test-set model selection
```

```python
from sklearn.model_selection import StratifiedGroupKFold, StratifiedKFold
idx = np.arange(len(df))

def run_cv(name, splits):
    P = np.zeros(len(df)); fold = np.zeros(len(df), int)
    for f, (tr, te) in enumerate(splits):
        shared = len(set(df.patient[tr]) & set(df.patient[te]))
        print(f"[{name}] fold {f+1}/5 | train {len(tr)} test {len(te)} | patients in BOTH: {shared}")
        P[te] = run_fold(tr, te, seed=f); fold[te] = f
    return P, fold

sp_patient = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(idx, y, groups=df.patient))
sp_image   = list(StratifiedKFold(5, shuffle=True, random_state=SEED).split(idx, y))

for tr, te in sp_patient:
    assert not set(df.patient[tr]) & set(df.patient[te])   # zero leakage

P_pat, F_pat = run_cv("patient-wise", sp_patient)
P_img, F_img = run_cv("image-level (leaky)", sp_image)
```

```python
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             roc_auc_score, confusion_matrix)

def metrics(yt, p):
    pr = (p > 0.5).astype(int)
    tn, fp, fn, tp = confusion_matrix(yt, pr, labels=[0, 1]).ravel()
    return dict(acc=accuracy_score(yt, pr), bal_acc=balanced_accuracy_score(yt, pr),
                f1=f1_score(yt, pr), auc=roc_auc_score(yt, p),
                sens=tp / (tp + fn), spec=tn / (tn + fp))

def table(P, F, name):
    out = []
    for mg in ["all", 40, 100, 200, 400]:
        m = np.ones(len(df), bool) if mg == "all" else (df.mag.values == mg)
        r = metrics(y[m], P[m])
        accs = [accuracy_score(y[m & (F == f)], (P[m & (F == f)] > .5).astype(int)) for f in range(5)]
        r.update(split=name, mag=mg, fold_std=np.std(accs))
        out.append(r)
    return pd.DataFrame(out)[["split", "mag", "acc", "fold_std", "bal_acc", "f1", "auc", "sens", "spec"]]

res = pd.concat([table(P_pat, F_pat, "patient-wise"), table(P_img, F_img, "image-level (leaky)")])
print(res.round(3).to_string(index=False))
res.to_csv("results.csv", index=False)

# Patient-level diagnosis: average probability over all of a patient's images
for name, P in [("patient-wise", P_pat), ("image-level (leaky)", P_img)]:
    g = pd.DataFrame(dict(patient=df.patient, p=P, y=y)).groupby("patient").agg(p=("p", "mean"), y=("y", "first"))
    print(f"{name}: patient-level accuracy = {accuracy_score(g.y, (g.p > .5).astype(int)):.3f}")
```

```python
def boot_ci(P, n=1000):
    d = pd.DataFrame(dict(patient=df.patient, mag=df.mag, ok=((P > .5).astype(int) == y).astype(int)))
    g = d.groupby(["patient", "mag"]).ok.agg(["sum", "size"]).reset_index()
    c = g.pivot(index="patient", columns="mag", values="sum").fillna(0).values
    t = g.pivot(index="patient", columns="mag", values="size").fillna(0).values
    rng = np.random.default_rng(0); N = len(c)
    bs = np.array([(c[s].sum(0) / t[s].sum(0)) for s in (rng.integers(0, N, N) for _ in range(n))])
    return np.percentile(bs, 2.5, 0), np.percentile(bs, 97.5, 0)

mags = [40, 100, 200, 400]
fig, ax = plt.subplots(1, 3, figsize=(17, 4.5))

for name, P, col in [("patient-wise", P_pat, "tab:blue"), ("image-level (leaky)", P_img, "tab:red")]:
    acc = [accuracy_score(y[df.mag == m], (P[df.mag == m] > .5).astype(int)) for m in mags]
    lo, hi = boot_ci(P)
    ax[0].errorbar(mags, acc, yerr=[np.array(acc) - lo, hi - np.array(acc)], marker="o", capsize=4, label=name, color=col)
ax[0].set_xticks(mags); ax[0].set_xlabel("Magnification"); ax[0].set_ylabel("Accuracy")
ax[0].set_title("Accuracy vs magnification (95% patient-bootstrap CI)"); ax[0].legend(); ax[0].grid(alpha=.3)

for name, P, col in [("patient-wise", P_pat, "tab:blue"), ("image-level (leaky)", P_img, "tab:red")]:
    from sklearn.metrics import roc_curve
    fpr, tpr, _ = roc_curve(y, P); ax[1].plot(fpr, tpr, color=col, label=f"{name} (AUC {roc_auc_score(y, P):.3f})")
ax[1].plot([0, 1], [0, 1], "k--", lw=.8); ax[1].set_title("ROC (all magnifications)")
ax[1].set_xlabel("FPR"); ax[1].set_ylabel("TPR"); ax[1].legend(); ax[1].grid(alpha=.3)

cm = confusion_matrix(y, (P_pat > .5).astype(int))
ax[2].imshow(cm, cmap="Blues")
for i in range(2):
    for j in range(2): ax[2].text(j, i, cm[i, j], ha="center", va="center", fontsize=14)
ax[2].set_xticks([0, 1], ["Benign", "Malignant"]); ax[2].set_yticks([0, 1], ["Benign", "Malignant"])
ax[2].set_title("Confusion matrix (patient-wise)"); ax[2].set_xlabel("Predicted"); ax[2].set_ylabel("True")
plt.tight_layout(); plt.savefig("results.png", dpi=150); plt.show()
```

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             roc_auc_score, roc_curve, confusion_matrix)

plt.rcParams.update({"font.size": 10, "axes.spines.top": False, "axes.spines.right": False,
                     "axes.titleweight": "bold", "axes.titlesize": 11})
C_PAT, C_IMG = "#1f5fa8", "#c0392b"
MAGS = [40, 100, 200, 400]
mag = df.mag.values
MODELS = {"Patient-wise (no leakage)": (P_pat, C_PAT), "Image-level (leakage)": (P_img, C_IMG)}

def boot_ci(P, n=2000, seed=0):
    d = pd.DataFrame(dict(patient=df.patient.values, mag=mag, ok=((P > .5).astype(int) == y).astype(int)))
    g = d.groupby(["patient", "mag"]).ok.agg(["sum", "size"]).reset_index()
    c = g.pivot(index="patient", columns="mag", values="sum").fillna(0).values
    t = g.pivot(index="patient", columns="mag", values="size").fillna(0).values
    rng = np.random.default_rng(seed); N = len(c)
    bs = np.array([c[s].sum(0) / t[s].sum(0) for s in (rng.integers(0, N, N) for _ in range(n))])
    return np.percentile(bs, 2.5, 0), np.percentile(bs, 97.5, 0)

def acc_by_mag(P):
    return np.array([accuracy_score(y[mag == m], (P[mag == m] > .5).astype(int)) for m in MAGS])

def overall(P):
    pr = (P > .5).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, pr).ravel()
    return [accuracy_score(y, pr), balanced_accuracy_score(y, pr), f1_score(y, pr),
            roc_auc_score(y, P), tp / (tp + fn), tn / (tn + fp)]

def cm_plot(ax, P, title, cmap):
    cm = confusion_matrix(y, (P > .5).astype(int)); cn = cm / cm.sum(1, keepdims=True)
    ax.imshow(cn, cmap=cmap, vmin=0, vmax=1)
    for i in range(2):
        for j in range(2):
            ax.text(j, i, f"{cn[i, j] * 100:.1f}%\n(n={cm[i, j]})", ha="center", va="center",
                    fontsize=11, color="white" if cn[i, j] > .5 else "black")
    ax.set_xticks([0, 1], ["Benign", "Malignant"]); ax.set_yticks([0, 1], ["Benign", "Malignant"])
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(title)
    for s in ax.spines.values(): s.set_visible(True)

def tag(ax, t): ax.text(-0.13, 1.06, t, transform=ax.transAxes, fontsize=13, fontweight="bold")

fig = plt.figure(figsize=(17, 10))
gs = GridSpec(2, 3, figure=fig, hspace=0.40, wspace=0.30)
fig.suptitle(f"BreakHis benign vs malignant | ResNet18, 5-fold CV | "
             f"{len(df):,} images, {df.patient.nunique()} patients", fontsize=14, fontweight="bold", y=0.97)

# (a) accuracy vs magnification with 95% patient-bootstrap CI
ax = fig.add_subplot(gs[0, 0]); tag(ax, "(a)")
ci_rows = []
for name, (P, col) in MODELS.items():
    a = acc_by_mag(P); lo, hi = boot_ci(P)
    ax.plot(MAGS, a, "-o", color=col, lw=2, label=name)
    ax.fill_between(MAGS, lo, hi, color=col, alpha=.15)
    for m, v in zip(MAGS, a): ax.annotate(f"{v:.3f}", (m, v), textcoords="offset points", xytext=(0, 8), ha="center", fontsize=8, color=col)
    ci_rows += [dict(model=name, mag=f"{m}x", acc=round(v, 3), ci_low=round(l, 3), ci_high=round(h, 3)) for m, v, l, h in zip(MAGS, a, lo, hi)]
ax.set_xticks(MAGS, [f"{m}×" for m in MAGS]); ax.set_ylim(0.76, 1.01)
ax.set_xlabel("Magnification"); ax.set_ylabel("Accuracy"); ax.set_title("Accuracy vs magnification (95% CI)")
ax.legend(frameon=False, loc="lower left"); ax.grid(alpha=.25)

# (b) ROC per magnification (patient-wise)
ax = fig.add_subplot(gs[0, 1]); tag(ax, "(b)")
for m, c in zip(MAGS, plt.cm.viridis(np.linspace(0.05, 0.85, 4))):
    k = mag == m; fpr, tpr, _ = roc_curve(y[k], P_pat[k])
    ax.plot(fpr, tpr, color=c, lw=1.8, label=f"{m}× (AUC {roc_auc_score(y[k], P_pat[k]):.3f})")
fpr, tpr, _ = roc_curve(y, P_pat)
ax.plot(fpr, tpr, "k--", lw=2, label=f"All (AUC {roc_auc_score(y, P_pat):.3f})")
ax.plot([0, 1], [0, 1], color="grey", lw=.8, ls=":")
ax.set_xlabel("False positive rate"); ax.set_ylabel("True positive rate")
ax.set_title("ROC by magnification (patient-wise)"); ax.legend(frameon=False, loc="lower right"); ax.grid(alpha=.25)

# (c) overall metrics comparison
ax = fig.add_subplot(gs[0, 2]); tag(ax, "(c)")
names = ["Accuracy", "Bal. acc.", "F1", "AUC", "Sensitivity", "Specificity"]
x = np.arange(len(names)); w = 0.38
for k, (name, (P, col)) in enumerate(MODELS.items()):
    v = overall(P); bars = ax.bar(x + (k - .5) * w, v, w, color=col, label=name)
    for b, val in zip(bars, v): ax.text(b.get_x() + b.get_width() / 2, val + .008, f"{val:.2f}", ha="center", fontsize=8)
ax.set_xticks(x, names, rotation=20); ax.set_ylim(0.5, 1.08); ax.set_ylabel("Score")
ax.set_title("Overall metrics"); ax.legend(frameon=False, loc="upper left", fontsize=8); ax.grid(axis="y", alpha=.25)

# (d, e) normalised confusion matrices
cm_plot(fig.add_subplot(gs[1, 0]), P_pat, "Confusion matrix: patient-wise", "Blues"); tag(plt.gca(), "(d)")
cm_plot(fig.add_subplot(gs[1, 1]), P_img, "Confusion matrix: image-level (leaky)", "Reds"); tag(plt.gca(), "(e)")

# (f) leakage gap per magnification
ax = fig.add_subplot(gs[1, 2]); tag(ax, "(f)")
gap = (acc_by_mag(P_img) - acc_by_mag(P_pat)) * 100
gap_all = (accuracy_score(y, (P_img > .5).astype(int)) - accuracy_score(y, (P_pat > .5).astype(int))) * 100
labels = [f"{m}×" for m in MAGS] + ["All"]; vals = list(gap) + [gap_all]
bars = ax.bar(labels, vals, color=["#7f8c8d"] * 4 + ["#2c3e50"])
for b, v in zip(bars, vals): ax.text(b.get_x() + b.get_width() / 2, v + .2, f"+{v:.1f}", ha="center", fontsize=9)
ax.set_ylabel("Accuracy inflation (percentage points)"); ax.set_title("Accuracy inflated by patient leakage")
ax.set_ylim(0, max(vals) * 1.2); ax.grid(axis="y", alpha=.25)

plt.savefig("metrics_report.png", dpi=300, bbox_inches="tight")
plt.savefig("metrics_report.pdf", bbox_inches="tight")
plt.show()

print(pd.DataFrame(ci_rows).to_string(index=False))
```
